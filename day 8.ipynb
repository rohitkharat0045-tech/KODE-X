```python
'''Exercise 1 - Function revision with real data logic

You have this employee data as a list of dicts:

employees = [
    {"name": "Alice",  "dept": "Engineering", "salary": 85000, "active": True},
    {"name": "Bob",    "dept": "Marketing",   "salary": 62000, "active": True},
    {"name": "Carol",  "dept": "Engineering", "salary": 92000, "active": True},
    {"name": "David",  "dept": "HR",          "salary": 55000, "active": False},
    {"name": "Eva",    "dept": "Marketing",   "salary": 78000, "active": False},
    {"name": "Frank",  "dept": "Engineering", "salary": 88000, "active": True},
    {"name": "Grace",  "dept": "Finance",     "salary": 67000, "active": True},
]

Write the following functions:

1. get_active_employees(employees)
   Returns list of only active employees.

2. average_salary_by_dept(employees)
   Returns dict: {dept: avg_salary}
   Round to 2 decimal places.

3. top_earner(employees)
   Returns the name and salary of the
   highest paid employee.

4. salary_band(employees)
   Returns a new list of dicts with an
   extra key "band" added to each employee:
   salary >= 80000 → "Senior"
   salary >= 60000 → "Mid"
   below 60000     → "Junior"

5. dept_headcount(employees)
   Returns dict: {dept: count}
   Only count active employees.

Call all functions and print results cleanly.
Use f-strings throughout.

Concepts: functions, list of dicts, comprehension,
          conditions, round(), f-string'''

employees = [
    {"name": "Alice",  "dept": "Engineering", "salary": 85000, "active": True},
    {"name": "Bob",    "dept": "Marketing",   "salary": 62000, "active": True},
    {"name": "Carol",  "dept": "Engineering", "salary": 92000, "active": True},
    {"name": "David",  "dept": "HR",          "salary": 55000, "active": False},
    {"name": "Eva",    "dept": "Marketing",   "salary": 78000, "active": False},
    {"name": "Frank",  "dept": "Engineering", "salary": 88000, "active": True},
    {"name": "Grace",  "dept": "Finance",     "salary": 67000, "active": True},
]

# 1. Active employees
def get_active_employees(employees):
    return [emp for emp in employees if emp["active"]]

# 2. Average salary by department
def average_salary_by_dept(employees):
    dept_salaries = {}
    for emp in employees:
        dept_salaries.setdefault(emp["dept"], []).append(emp["salary"])
    return {dept: round(sum(sals)/len(sals), 2) for dept, sals in dept_salaries.items()}

# 3. Top earner
def top_earner(employees):
    top = max(employees, key=lambda e: e["salary"])
    return {"name": top["name"], "salary": top["salary"]}

# 4. Salary band
def salary_band(employees):
    banded = []
    for emp in employees:
        if emp["salary"] >= 80000:
            band = "Senior"
        elif emp["salary"] >= 60000:
            band = "Mid"
        else:
            band = "Junior"
        banded.append({**emp, "band": band})
    return banded

# 5. Department headcount (active only)
def dept_headcount(employees):
    counts = {}
    for emp in employees:
        if emp["active"]:
            counts[emp["dept"]] = counts.get(emp["dept"], 0) + 1
    return counts


# ---- Calling functions ----
active = get_active_employees(employees)
avg_salary = average_salary_by_dept(employees)
top = top_earner(employees)
banded = salary_band(employees)
headcount = dept_headcount(employees)

print("\n--- Active Employees ---")
for emp in active:
    print(f"{emp['name']} ({emp['dept']}, Salary: {emp['salary']})")

print("\n--- Average Salary by Dept ---")
for dept, avg in avg_salary.items():
    print(f"{dept}: {avg}")

print("\n--- Top Earner ---")
print(f"{top['name']} with salary {top['salary']}")

print("\n--- Employees with Salary Band ---")
for emp in banded:
    print(f"{emp['name']} ({emp['dept']}, {emp['salary']}) → {emp['band']}")

print("\n--- Department Headcount (Active Only) ---")
for dept, count in headcount.items():
    print(f"{dept}: {count}")
```

```python
'''Exercise 2 - Exception Handling

Write a function called safe_divide(a, b) that:
- Returns a divided by b
- Handles ZeroDivisionError gracefully
- Handles TypeError if non-numbers are passed
- Logs a message for each error type
- Returns None on error instead of crashing

Then write a function called process_salaries(data)
that takes a list like:
[85000, 0, "N/A", 62000, None, 78000]

For each value, try dividing 1200000 by it
(simulating yearly-to-monthly).
Use safe_divide and print either the result
or the error reason.

Concepts: try, except, ZeroDivisionError,
          TypeError, None, logging or print
'''
# 1. Safe divide function
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print(f"Error: Cannot divide {a} by zero.")
        return None
    except TypeError:
        print(f"Error: Invalid types for division ({a} / {b}).")
        return None


# 2. Process salaries function
def process_salaries(data):
    for val in data:
        result = safe_divide(1200000, val)
        if result is not None:
            print(f"1200000 / {val} = {result:.2f}")
        else:
            print(f"Skipping value: {val}")


# ---- Test run ----
salaries = [85000, 0, "N/A", 62000, None, 78000]

process_salaries(salaries)
```

```python
'''Exercise 3 - File Handling revision: JSON

You are given this JSON string (paste it
into your code as a variable or save as
employees.json):

{
  "company": "KloudOcean",
  "employees": [
    {"id": 1, "name": "Alice",  "dept": "Engineering", "salary": 85000},
    {"id": 2, "name": "Bob",    "dept": "Marketing",   "salary": 62000},
    {"id": 3, "name": "Carol",  "dept": "Engineering", "salary": 92000},
    {"id": 4, "name": "David",  "dept": "HR",          "salary": 55000},
    {"id": 5, "name": "Grace",  "dept": "Finance",     "salary": 67000}
  ]
}

Write code to:
1. Parse the JSON using the json module.
2. Print the company name.
3. Print all employee names and salaries.
4. Find and print the highest paid employee.
5. Add a new employee to the list:
   id 6, name "Henry", dept "Finance",
   salary 71000.
6. Write the updated data back to employees.json
   with indent=2.
7. Read and verify the file was written correctly.

Concepts: json.loads(), json.dumps(),
          json.load(), json.dump(), indent


---------------------------------------------
'''

import json

# JSON string (could also be saved in employees.json)
json_data = """
{
  "company": "KloudOcean",
  "employees": [
    {"id": 1, "name": "Alice",  "dept": "Engineering", "salary": 85000},
    {"id": 2, "name": "Bob",    "dept": "Marketing",   "salary": 62000},
    {"id": 3, "name": "Carol",  "dept": "Engineering", "salary": 92000},
    {"id": 4, "name": "David",  "dept": "HR",          "salary": 55000},
    {"id": 5, "name": "Grace",  "dept": "Finance",     "salary": 67000}
  ]
}
"""

# 1. Parse JSON string
data = json.loads(json_data)

# 2. Print company name
print(f"Company: {data['company']}")

# 3. Print all employee names and salaries
print("\n--- Employees ---")
for i in data["employees"]:
    print(f"{i['name']} → Salary: {i['salary']}")

# 4. Find highest paid employee
top = max(data["employees"], key=lambda e: e["salary"])
print(f"\nTop Earner: {top['name']} with salary {top['salary']}")

# 5. Add new employee
new_emp = {"id": 6, "name": "Henry", "dept": "Finance", "salary": 71000}
data["employees"].append(new_emp)

# 6. Write updated data back to employees.json
with open("employees.json", "w") as f:
    json.dump(data, f, indent=2)

print("\nUpdated data written to employees.json")

# 7. Read and verify file
with open("employees.json", "r") as f:
    verified = json.load(f)

print("\n--- Verified Employees ---")
for i in verified["employees"]:
    print(f"{i['id']}: {i['name']} ({i['dept']}) → {i['salary']}")
```

```python
'''Exercise 5 - Pandas: DataFrame Operations

Using the same patients.csv:

1. Add a new column called fee_category:
   fee >= 700 → "High"
   fee >= 400 → "Medium"
   below 400  → "Low"
   Use np.select() or apply() with a function.

2. Add a new column called age_group:
   age < 30   → "Young"
   age < 50   → "Adult"
   age >= 50  → "Senior"

3. Find the average fee per city.
   Use groupby().

4. Count patients per blood_type.
   Use value_counts().

5. Find the city with the highest average fee.

6. Export the updated DataFrame to a new file:
   patients_processed.csv
   Do not include the index.

Concepts: apply(), lambda, np.select(),
          groupby(), mean(), value_counts(),
          to_csv(index=False)
'''
import pandas as pd


# 1. Read CSV
df = pd.read_csv("patients.csv")

# 1. Add fee_category column
conditions = [
    (df["fee"] >= 700),
    (df["fee"] >= 400) & (df["fee"] < 700),
    (df["fee"] < 400)
]
choices = ["High", "Medium", "Low"]

# Alternative with apply + lambda:
# df["fee_category"] = df["fee"].apply(lambda x: "High" if x >= 700 else "Medium" if x >= 400 else "Low")

# 2. Add age_group column
def age_group(age):
    if age < 30:
        return "Young"
    elif age < 50:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)

# 3. Average fee per city
avg_fee_city = df.groupby("city")["fee"].mean()
print("\n--- Average Fee per City ---")
print(avg_fee_city)

# 4. Count patients per blood_type
blood_counts = df["blood_type"].value_counts()
print("\n--- Patients per Blood Type ---")
print(blood_counts)

# 5. City with highest average fee
top_city = avg_fee_city.idxmax()
top_fee = avg_fee_city.max()
print(f"\nCity with highest average fee: {top_city} ({top_fee})")

# 6. Export updated DataFrame
df.to_csv("patients_processed.csv", index=False)
print("\npatients_processed.csv written successfully!")
```
