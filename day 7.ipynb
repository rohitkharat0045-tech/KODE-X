```python
'''━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Exercise 1 - Tuples

Create a tuple storing details of one employee:
(101, "Priya Mehta", "IT", 85000, "Pune")

Write code to:
1. Access name and salary using index.
2. Slice the tuple to get only the last 3 values.
3. Try to change the salary value.
   Observe the error. Write a comment explaining
   why this happens.
4. Unpack all 5 values into separate variables.
   Print each variable with a label.
5. Check if "IT" exists in the tuple.
6. Count how many times "Pune" appears in this tuple:
   ("Pune", "Mumbai", "Pune", "Delhi", "Pune")
7. Find the index of "Mumbai" in the above tuple.

Concepts: indexing, slicing, immutability,
          unpacking, in, count(), index()
'''

# Create the tuple
employee = (101, "Priya Mehta", "IT", 85000, "Pune")

# 1. Access name and salary using index
print("Name:", employee[1])
print("Salary:", employee[3])

# 2. Slice the tuple to get last 3 values
print("Last 3 values:", employee[-3:])

# 3. Try to change the salary value
'in tuple the values are not change beacuse tuple is immutable'

# 4. Unpack all 5 values
emp_id, name, dept, salary, city = employee
print("ID:", emp_id)
print("Name:", name)
print("Department:", dept)
print("Salary:", salary)
print("City:", city)

# 5. Check if "IT" exists in the tuple
print("Is 'IT' in tuple?", "IT" in employee)

# 6. Count how many times "Pune" appears
cities = ("Pune", "Mumbai", "Pune", "Delhi", "Pune")
print("Count of Pune:", cities.count("Pune"))

# 7. Find the index of "Mumbai"
print("Index of Mumbai:", cities.index("Mumbai"))
```

```python
'''Exercise 2 - Sets

Create two sets:
batch_a = {"Rahul", "Priya", "Amit", "Neha", "Karan"}
batch_b = {"Priya", "Karan", "Sneha", "Divya", "Rohan"}

Write code to:
1. Find students present in both batches.
   (Intersection)
2. Find all unique students across both batches.
   (Union)
3. Find students only in batch_a, not in batch_b.
   (Difference)
4. Find students who are in one batch but not both.
   (Symmetric difference)
5. Check if {"Priya", "Karan"} is a subset of batch_a.
6. Add "Meera" to batch_a.
7. Remove "Amit" from batch_a safely.
   Use discard() so no error if name is missing.
8. Convert batch_a to a sorted list and print it.

Concepts: intersection, union, difference,
          symmetric_difference, issubset,
          add(), discard(), sorted()
'''

# Create two sets
batch_a = {"Rahul", "Priya", "Amit", "Neha", "Karan"}
batch_b = {"Priya", "Karan", "Sneha", "Divya", "Rohan"}

# 1. Intersection → students in both batches
print("Both batches:", batch_a & batch_b)

# 2. Union → all unique students
print("All students:", batch_a | batch_b)

# 3. Difference → only in batch_a
print("Only in batch_a:", batch_a - batch_b)

# 4. Symmetric difference → in one batch but not both
print("One batch only:", batch_a ^ batch_b)

# 5. Subset check
print("Is {'Priya','Karan'} subset of batch_a?", {"Priya", "Karan"}.issubset(batch_a))

# 6. Add "Meera" to batch_a
batch_a.add("Meera")
print("After adding Meera:", batch_a)

#7. Remove "Amit" from batch_a safely.Use discard() so no error if name is missing.
batch_a.discard("Amit")
print("After discarding Amit:", batch_a)

# 8. Convert to sorted list
sorted_list = sorted(batch_a)
print("Sorted list:", sorted_list)
```

```python
'''Exercise 3 - Functions Basics

Write the following functions:

1. greet(name)
   Returns "Hello, Priya! Welcome to Data Engineering."
   Use f-string inside the function.

2. calculate_tax(salary)
   If salary > 50000, tax is 20 percent.
   If salary between 30000 and 50000, tax is 10 percent.
   Below 30000, no tax.
   Return the tax amount.

3. is_palindrome(word)
   Returns True if the word reads the same forwards
   and backwards. Ignore case.
   Test with: "Madam", "Python", "Racecar"

4. count_vowels(sentence)
   Returns the count of vowels in the sentence.

Call all four functions and print results
with clear labels.

Concepts: def, return, f-string, condition,
          slicing, lower(), loop'''

# 1. Greet function
def greet(name):
    return f"Hello, {name}! Welcome to Data Engineering."

# 2. Calculate tax
def calculate_tax(salary):
    if salary > 50000:
        return salary * 0.20
    elif salary >= 30000:
        return salary * 0.10
    else:
        return 0

# 3. Palindrome check
def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]

# 4. Count vowels
def count_vowels(sentence):
    vowels = "aeiou"
    return sum(1 for ch in sentence.lower() if ch in vowels)

#---output/printing function

# 1. Greeting
print(greet("Priya"))

# 2. Tax calculation
print("Tax on 60000:", calculate_tax(60000))
print("Tax on 40000:", calculate_tax(40000))
print("Tax on 20000:", calculate_tax(20000))

# 3. Palindrome tests
print("Is 'Madam' palindrome?", is_palindrome("Madam"))
print("Is 'Python' palindrome?", is_palindrome("Python"))
print("Is 'Racecar' palindrome?", is_palindrome("Racecar"))

# 4. Vowel count
sentence = "Data Engineering with Python"
print(f"Vowel count in '{sentence}':", count_vowels(sentence))
```

```python
'''
Exercise 4 - Functions with Default Arguments

Write a function called employee_summary that takes:
- name (required)
- department (required)
- salary (default value: 50000)
- city (default value: "Pune")

The function prints a formatted employee card.

Call the function four times:
1. Passing all four arguments.
2. Passing only name and department.
3. Passing name, department, and a custom salary.
4. Passing name, department, and a custom city.

Observe how default arguments work in each case.

Concepts: def, default arguments, f-string
'''

# Function with default arguments
def employee_summary(name, department, salary=50000, city="Pune"):
    print(f"--- Employee Card ---")
    print(f"Name: {name}")
    print(f"Department: {department}")
    print(f"Salary: {salary}")
    print(f"City: {city}")
    print("---------------------\n")


# 1. Passing all four arguments
employee_summary("raja kumara", "ITI", 85000, "Mumbai")

# 2. Passing only name and department
employee_summary("rohit Sharma", "crickter")

# 3. Passing name, department, and custom salary
employee_summary("Sahiba bali", "HR", 60000)

# 4. Passing name, department, and custom city
employee_summary("Amit mishra", "Engineering",   city="Delhi")
```

```python
'''Exercise 5 - *args and **kwargs

Part A - *args

Write a function called total_sales(*args) that:
- Accepts any number of sales figures
- Returns the total and average

Test with:
total_sales(1500, 2300, 4100)
total_sales(800, 1200, 3300, 2700, 4500)

Part B - **kwargs

Write a function called create_profile(**kwargs) that:
- Accepts any number of keyword arguments
- Prints each key and value on a new line

Test with:
create_profile(name="Rahul", city="Pune",
               role="Data Engineer", salary=72000)

Part C - Combined

Write a function called log_event(event_type,
*args, **kwargs) that:
- Prints the event type
- Prints all positional values as a list
- Prints all keyword values as key:value pairs

Test with:
log_event("pipeline_run", "step1", "step2",
          status="success", rows=5000)

Concepts: *args, **kwargs, loop, mixed arguments
'''
def print_seprator():
    print("====================================================")
def total_sales(*args):
    total = sum(args)
    average = total / len(args) if args else 0
    return total, average
print(total_sales(1500, 2300, 4100))  
print(total_sales(800, 1200, 3300, 2700, 4500))
print_seprator()

def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

create_profile(name="Rahul", city="Pune",
               role="Data Engineer", salary=72000)
print_seprator()

def log_event(event_type, *args, **kwargs):
    print(f"Event Type: {event_type}")
    print("Positional Values:", list(args))
    print("Keyword Values:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

log_event("pipeline_run", "step1", "step2",
          status="success", rows=5000)
print_seprator()
```

```python
'''Exercise 6 - List Comprehensions

Given this list of salaries:
salaries = [72000, 48000, 85000, 61000, 55000,
            52000, 47000, 91000, 38000, 67000]

Using list comprehensions (one line each):
1. Create a list of salaries above 60000.
2. Create a list of salaries after applying
   10 percent increment to each.
3. Create a list of "High" or "Low" for each salary
   based on whether it is above 60000 or not.
4. Create a list of squares of all salaries
   below 50000.

Then given this list of names:
names = ["rahul", "AMIT", "Priya", "NEHA", "karan"]

5. Create a new list with all names in Title Case.
6. Create a new list of names that have
   more than 4 characters.

Concepts: list comprehension, condition inside
          comprehension, string methods
'''

salaries = [72000, 48000, 85000, 61000, 55000,52000, 47000, 91000, 38000, 67000]

#1. Create a list of salaries above 60000.
abvoe60000_salaries=[i for i in salaries if i> 60000]
print(abvoe60000_salaries)

#2. Create a list of salaries after applying  10 percent increment to each.

after10_increment=[i *1.10 for i in salaries]
print(after10_increment)

#4. Create a list of squares of all salaries below 50000.

squares=[i**2 for i in salaries if i < 50000]
print(squares)

#3. Create a list of "High" or "Low" for each salary based on whether it is above 60000 or not

labels = ["High" if s > 60000 else "Low" for s in salaries]
print(labels)

print("============================================================================")

names = ["rahul", "AMIT", "Priya", "NEHA", "karan"]

#5. Create a new list with all names in Title Case.

title_case = [n.title() for n in names]
print(title_case)


#6. Create a new list of names that have more than 4 characters.

long_names = [n for n in names if len(n) > 4]
print(long_names)
```

```python
'''Exercise 7 - Dict Comprehensions

Part A
Given two lists:
names   = ["Rahul", "Priya", "Amit", "Neha", "Karan"]
salaries = [72000, 85000, 48000, 61000, 55000]

Create a dictionary {name: salary} using
dict comprehension.

Part B
From the dictionary created above:
1. Create a new dict with only employees
   earning above 60000.
2. Create a new dict where each salary
   is increased by 5000.
3. Create a new dict where keys are names
   and values are "High" or "Low" based
   on salary threshold of 60000.

All in one line each using dict comprehension.

Concepts: dict comprehension, zip(),
          condition, arithmetic'''

#PART A
names   = ["Rahul", "Priya", "Amit", "Neha", "Karan"]
salaries = [72000, 85000, 48000, 61000, 55000]

emp_dict = {name:i for name, i in zip(names, salaries)}
print(emp_dict)

print("----------------------------=================--------------------------")

#PART B
#1. Create a new dict with only employeesearning above 60000.
high_saliery = {name: i for name, i in emp_dict.items() if i > 60000}
print(high_saliery)
print("----------------------------=================--------------------------")

#2. Create a new dict where each salaryis increased by 5000.
incremented = {name: i+ 5000 for name, i in emp_dict.items()}
print(incremented)
print("----------------------------=================--------------------------")

#3. Create a new dict where keys are namesand values are "High" or "Low" basedon salary threshold of 60000.
position= {name: ("High" if i > 60000 else "Low") for name, i in emp_dict.items()}
print(position)
print("----------------------------=================--------------------------")
```
