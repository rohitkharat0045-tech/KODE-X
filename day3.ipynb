```python
'''Exercise 1 – Sum of Even Numbers
Given a list:
numbers = [3, 8, 12, 5, 7, 10]
Write a program to calculate the sum of only the even numbers in the list.
Expected Output Example:
Sum of even numbers: 30
'''
numbers = [3, 8, 12, 5, 7, 10]

sum_even = 0
for num in numbers:
    if num % 2 == 0:
        sum_even += num

print("Sum of even numbers:", sum_even)
```

```python
'''Exercise 2 – Count Vowels in a String

Ask the user to enter a word.

Count how many vowels are present in the word.

Example:

Input: education
Output: Number of vowels = 5

Hint: vowels = a, e, i, o, u

---'''
word = input("Enter a word: ")
vowels = "aeiou"
count = 0

for i in word.lower():
    if i in vowels:
        count += 1

print("Number of vowels =", count)
```

```python
numbers = [45, 12, 78, 3, 25]

minimum = numbers[0]
for num in numbers:
    if num < minimum:
        minimum = num

print("Smallest number:", minimum)
```

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for num in numbers:
    squares.append(num ** 2)

print(squares)
```

```python
word = input("Enter a word: ")
freq = {}

for ch in word:
    freq[ch] = freq.get(ch, 0) + 1

for key, value in freq.items():
    print(f"{key} : {value}")
```
