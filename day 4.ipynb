```python
'''Exercise 2 – Function to Check Even or Odd

Write a function called check_even_odd that takes a number as input.

If the number is even print “Even”.
If the number is odd print “Odd”.
'''
def check_even_odd(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

check_even_odd(10)   
check_even_odd(7)    
```

```python
'''Exercise 3 – Function to Find Largest Number

Write a function that takes a list of numbers and returns the largest number.

Example list:

[10, 45, 2, 78, 33]
'''

def find_largest_number(num):
    largest = num[0]  
    for i in num:
        if i > largest:
            largest = i
    return largest

nums = [10, 45, 2, 78, 33]
print(f"largest number: {find_largest_number(nums)}")
```

```python
'''Exercise 4 – Count Words in a Sentence

Ask the user to enter a sentence.

Write a function that counts how many words are present in the sentence.

Example

Input
Python is easy to learn

Output
Number of words: 5'''


def count_words(sentence):
    words = sentence.split()   
    return len(words)
user_input = input("Enter a sentence: ")
print("Number of words:", count_words(user_input))

```

```python
'''Exercise 6 – Read Data from a File

Read the file students.txt created earlier.

Print each name line by line.

Concepts to use

open()
readlines()
loop
'''
with open("students.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())
```

```python
'''Exercise 10 – Find Longest Word

Ask the user to enter a sentence.

Write a function that finds the longest word in the sentence.

Example

Input
Python programming is powerful

Output
Longest word: programming
'''
def find_longest_word(sentence):
    words = sentence.split()
    longest = ""   

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

sentence = input("Enter a sentence: ")
result = find_longest_word(sentence)
print(f"Longest word: {result}")
```
