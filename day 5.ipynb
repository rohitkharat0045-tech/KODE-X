```python
'''Exercise 1  String Length Without len()

Given a string:

text = “dataengineering”

Find the length of the string without using the len() function.

Hint: Use a loop.

⸻'''

text = "dataengineering"

count = 0
for i in text:   
    count += 1      
print("Length of string:", count)
```

```python
'''Exercise 3 Count Occurrence of a Character

Ask the user to enter a word.

Ask the user to enter a character.

Count how many times that character appears in the word.

Example

Word: banana
Character: a'''


word = input("Enter a word: ")
char = input("Enter a character: ")
count = 0
for c in word:
    if c == char:  
        count += 1

print(f"The character '{char}' appears {count} times in the word '{word}'.")
```

```python
'''Exercise 6 Dictionary Creation Using Loop

Create an empty dictionary.

Ask the user to enter names of 3 students and their marks.

Store them in a dictionary.

Example

Rahul 85
Amit 72
Priya 90

Print the dictionary.'''


students = {}

for i in range(3):
    name = input("Enter student name: ")
    marks = int(input(f"Enter marks for {name}: "))
    students[name] = marks 
print("Student Dictionary:", students)
```

```python
'''Exercise 7 Find Student With Highest Marks

Given dictionary:

marks = {
Rahul:85,
Amit:72,
Priya:90,
Neha:78
}

Find the student who scored the highest marks.'''

marks = {
    "Rahul": 85,
    "Amit": 72,
    "Tanvi": 90,
    "Neha": 78
}
highest_student = max(marks, key=marks.get)
highest_marks = marks[highest_student]
print(f"The student with highest marks is {highest_student} ({highest_marks}).")
```

```python
'''Exercise 8  Function to Count Words

Write a function called count_words.

The function should take a sentence as input and return the number of words.

Example

Input
Python is widely used in data engineering

Output
7'''
def count_words(sentence):
    words = sentence.split()
    return len(words)
text = "Python is widely used in data engineering"
print(count_words(text),"THALA FOR REASON")
```

```python
'''
Exercise 9  Read File and Print Line Numbers

Create a file called notes.txt.

Add at least 5 lines of text.

Write a Python program to read the file and print line numbers with each line.

Example output

1 Python is powerful
2 Data engineering uses Python
3 Linux is important'''

with open("notes.txt", "w") as file:
    file.write("Python is powerful\n")
    file.write("Data engineering uses Python\n")
    file.write("Linux is important\n")
  
with open("notes.txt", "r") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"{line_number} {line.strip()}")
```

```python
'''Exercise 10 Filter Data From File

Create a file called scores.txt.

Example data:

Rahul 85
Amit 60
Priya 92
Neha 55
Karan 75

Write a Python program to display only students whose marks are greater than 70.
'''
with open("scores.txt", "w") as file:
    file.write("Rahul 85\n")
    file.write("Amit 60\n")
    file.write("Priya 92\n")
    file.write("Neha 55\n")
    file.write("Karan 75\n")

with open("scores.txt", "r") as file:
    for line in file:
        name, marks = line.split()
        marks = int(marks)  
        if marks > 70:
            print(f"{name} {marks}")
```

```python
'''Exercise 11  CSV File Reading

Create a CSV file called employees.csv.

Example structure:

id,name,salary

1,Rahul,50000
2,Amit,45000
3,Priya,70000

Write a Python program to read and print each row.

Hint

Use csv module.
'''

import csv

with open("employees.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["id", "name", "salary"])
    writer.writerow([1, "Rahul", 50000])
    writer.writerow([2, "Amit", 45000])
    writer.writerow([3, "Priya", 70000])

with open("employees.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
