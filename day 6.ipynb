```python
'=====================================Exercise 2 - List Operations================='

#Given this list of sales figures:
sales=[1500,4200,3100,870,5600,2200,4800,990]
max_sales=sales[0]

#1. Find the maximum value using a loop.
for i in sales:
    if i > max_sales:
        max_sales=i
print(f"maximum sales:",max_sales)
```

```python
#2. Find the minimum value using a loop.
min_sales=sales[0]
for i in sales:
    if i < min_sales:
        min_sales=i


print(f"minimum sales:",min_sales)         
```

```python
#3. Calculate the total and average.
total_sales=sales[0]
for  i in sales:
    total_sales+=i
average_sales=total_sales/len(sales)  
print(f"total sales:",total_sales)
print(f"average sales:",average_sales)  
```

```python
#4. Count how many values are above 3000.
count=0
for i in sales:
    if i  > 3000:
        count +=1
print(f"the values are above 3000 are:",count)        
```

```python
#5. Create a new list containing only values above 3000.
new_list=[]
for i in sales:
    if i > 3000:
        new_list.append(i)
print(f"new list:",new_list)
```

```python
#5. Count students per department.
students = [
    ["Rahul", "IT", 85],
    ["Amit", "HR", 62],
    ["Priya", "IT", 91],
    ["Neha", "Finance", 74],
    ["Karan", "IT", 58],
    ["Sneha", "HR", 88]
]
dept_count = {}

for student in students:
    dept = student[1]   
    if dept in dept_count:
        dept_count[dept] += 1
    else:
        dept_count[dept] = 1

for dept, count in dept_count.items():
    print(f"{dept:<8} : {count}")        
```

```python
'====================================Exercise 4 - String Methods Practice================================================================'
#1. Remove leading and trailing spaces using strip().
raw="rahul sharma,data enginner, pune , 85000  "
print(raw.strip( ))

#2. Split the string by comma into a list.
print(raw.split(','))

#Strip spaces from each item in the list.
space=[item.strip() for item in raw.split(',')]
print(space)

#4. Convert name to Title Case.
space=space[0].title()
print(space)
```

```python
'=========================================Exercise 5 - String Search and Replace====================================================='
'======================================='

text = "Data engineering involves data ingestion,data transformation, and data storage."
#1. Count how many times the word "data" appears  Case-insensitive count.

print(text.count('data') )

count=text.lower().count('data')
print(count)
#- Replace every occurrence of "data" with "DATA".

replaced = text.replace("data", "DATA").replace("Data", "DATA")
print(replaced)

#Check if the sentence starts with "Data".
print(text.startswith("Data"))

# 4. Check if the sentence ends with "storage."
print(text.endswith("storage."))

# 5. Find the index position of the first occurrence of "transformation"
index = text.find("transformation")
print(index)

# 6. Extract the word "transformation" using slicing
word = text[index:index+len("transformation")]
print(word)
```

```python
'===================Exercise 6 - Dictionary Basics=============='

employee = {
    "name": "Rohit Kharat",
    "department": "Data Engineering",
    "salary": 45000,
    "city": "Pune",
    "is_active": True
}
# 1. Print all keys
print(employee.keys())

# 2. Print all values
print(employee.values())

# 3. Print all key-value pairs using items()
print(employee.items())

# 4. Update the salary to a new value
employee["salary"] = 90000
print("Updated salary:", employee["salary"])

# 5. Add a new key: years_of_experience with value 3
employee["years_of_experience"] = 3
print("Added years_of_experience:", employee["years_of_experience"])

# 6. Delete the is_active key
del employee["is_active"]
print("After deleting is_active:", employee)

# 7. Check if the key "department" exists
print("department" in employee)

# 8. Get the value of a key that does not exist using .get()
print(employee.get("designation", "Not Found"))



```

```python
'==============Exercise 7 - Dictionary from Two Lists========='
columns = ["emp_id", "name", "department", "salary", "city"]
values  = [101, "Priya Mehta", "IT", 85000, "Pune"]

# 1. Create a dictionary by combining both lists using zip()
new_dict = dict(zip(columns, values))

# 2. Print the dictionary
print(new_dict)

# 3. Print only the name and salary
print("Name:", new_dict["name"])
print("Salary:", new_dict["salary"])

# 4. Convert the salary from integer to a formatted string
formatted_salary = f"Rs. {new_dict['salary']:,}"
print("Formatted Salary:", formatted_salary)
```
