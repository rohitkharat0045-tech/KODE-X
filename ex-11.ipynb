```python
'''━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
PYTHON REVISION
Topics: Functions, *args, **kwargs,
        Pandas, map, filter, lambda
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Exercise 1 - Functions + *args + **kwargs

You have this data as a list of dicts:

employees = [
    {"name": "Alice",  "dept": "Engineering",
     "salary": 85000, "years": 5},
    {"name": "Bob",    "dept": "Marketing",
     "salary": 62000, "years": 3},
    {"name": "Carol",  "dept": "Engineering",
     "salary": 92000, "years": 8},
    {"name": "David",  "dept": "HR",
     "salary": 55000, "years": 2},
    {"name": "Eva",    "dept": "Finance",
     "salary": 78000, "years": 6},
    {"name": "Frank",  "dept": "Engineering",
     "salary": 88000, "years": 7},
    {"name": "Grace",  "dept": "Finance",
     "salary": 67000, "years": 4},
]

Write these three functions:

1. dept_summary(employees, dept)
   Returns a dict with keys:
   count, avg_salary, max_salary, min_salary
   for that department only.

2. apply_increment(employees,
                   *dept_names,
                   **increment_rates)
   For each employee:
   - If their dept is in dept_names AND
     a rate exists in increment_rates
     for that dept, apply that rate.
   - Otherwise apply default 5 percent.
   Return a NEW list. Do not modify original.

   Test with:
   apply_increment(
       employees,
       "Engineering", "Finance",
       Engineering=0.15,
       Finance=0.10
   )

3. print_payslip(employee, **options)
   Options:
     currency   (default "Rs.")
     show_tax   (default True)
     tax_rate   (default 0.10)
   Prints a formatted payslip.

Call all three. Print results with labels.

Concepts: functions, *args, **kwargs,
          list of dicts, .get(), round()
'''

employees = [
    {"name": "Alice",  "dept": "Engineering", "salary": 85000, "years": 5},
    {"name": "Bob",    "dept": "Marketing",   "salary": 62000, "years": 3},
    {"name": "Carol",  "dept": "Engineering", "salary": 92000, "years": 8},
    {"name": "David",  "dept": "HR",          "salary": 55000, "years": 2},
    {"name": "Eva",    "dept": "Finance",     "salary": 78000, "years": 6},
    {"name": "Frank",  "dept": "Engineering", "salary": 88000, "years": 7},
    {"name": "Grace",  "dept": "Finance",     "salary": 67000, "years": 4},
]

# 1. Department summary
def dept_summary(employees, dept):
    dept_emps = [e for e in employees if e["dept"] == dept]
    salaries = [e["salary"] for e in dept_emps]
    return {
        "count": len(dept_emps),
        "avg_salary": round(sum(salaries) / len(salaries), 2) if salaries else 0,
        "max_salary": max(salaries) if salaries else None,
        "min_salary": min(salaries) if salaries else None,
    }

# 2. Apply increment
def apply_increment(employees, *dept_names, **increment_rates):
    new_list = []
    for e in employees:
        rate = increment_rates.get(e["dept"], 0.05)  # default 5%
        if e["dept"] in dept_names:
            rate = increment_rates.get(e["dept"], rate)
        new_salary = round(e["salary"] * (1 + rate), 2)
        new_list.append({**e, "salary": new_salary})
    return new_list

# 3. Print payslip
def print_payslip(employee, **options):
    currency = options.get("currency", "Rs.")
    show_tax = options.get("show_tax", True)
    tax_rate = options.get("tax_rate", 0.10)

    gross = employee["salary"]
    tax = gross * tax_rate if show_tax else 0
    net = gross - tax

    print(f"Payslip for {employee['name']}")
    print(f"Department: {employee['dept']}")
    print(f"Gross Salary: {currency} {gross:,.2f}")
    if show_tax:
        print(f"Tax ({tax_rate*100:.0f}%): {currency} {tax:,.2f}")
    print(f"Net Salary: {currency} {net:,.2f}")
    print("-" * 30)

# -------------------------
# Call all three functions
# -------------------------

print("Department Summary (Engineering):")
print(dept_summary(employees, "Engineering"))

print("\nIncremented Salaries:")
updated = apply_increment(
    employees,
    "Engineering", "Finance",
    Engineering=0.15,
    Finance=0.10
)
for e in updated:
    print(e)

print("\nPayslip Example:")
print_payslip(updated[0], currency="Rs.", show_tax=True, tax_rate=0.10)
```

```python
'''Exercise 2 - Pandas revision

Create employees.csv from the data above.
Add a column: active (True/False alternating).
Load into a DataFrame.

1. Filter employees where salary is above
   the mean salary of their own department.
   Use groupby().transform('mean'):
   df[df['salary'] > df.groupby('dept')
      ['salary'].transform('mean')]

2. Add experience_band column:
   years >= 7 → "Veteran"
   years >= 4 → "Experienced"
   below 4    → "Fresher"
   Use apply() or pd.cut().

3. Find the department with the largest
   gap between max and min salary.
   Use groupby().agg() then subtract.

4. For each department find the name of
   the highest paid employee.
   Use groupby().idxmax() on salary,
   then use that index to look up the name.

5. Build a summary DataFrame:
   dept | headcount | avg_salary | top_earner
   where top_earner is the name of the
   highest paid person per department.

Concepts: transform, apply, idxmax,
          groupby().agg(), pd.cut()


---------------------------------------------
'''

import pandas as pd

# Step 1: Create employees.csv (only needed once)
employees = [
    {"name": "Alice",  "dept": "Engineering", "salary": 85000, "years": 5},
    {"name": "Bob",    "dept": "Marketing",   "salary": 62000, "years": 3},
    {"name": "Carol",  "dept": "Engineering", "salary": 92000, "years": 8},
    {"name": "David",  "dept": "HR",          "salary": 55000, "years": 2},
    {"name": "Eva",    "dept": "Finance",     "salary": 78000, "years": 6},
    {"name": "Frank",  "dept": "Engineering", "salary": 88000, "years": 7},
    {"name": "Grace",  "dept": "Finance",     "salary": 67000, "years": 4},
]

# Add alternating active column
for i, e in enumerate(employees):
    e["active"] = (i % 2 == 0)

pd.DataFrame(employees).to_csv("employees.csv", index=False)

# Step 2: Load into DataFrame
df = pd.read_csv("employees.csv")

# 1. Filter employees with salary above dept mean
above_mean = df[df["salary"] > df.groupby("dept")["salary"].transform("mean")]
print("Employees above dept mean salary:\n", above_mean, "\n")

# 2. Add experience_band column
df["experience_band"] = pd.cut(
    df["years"],
    bins=[-1, 3, 6, float("inf")],
    labels=["Fresher", "Experienced", "Veteran"]
)
print("With experience_band:\n", df, "\n")

# 3. Dept with largest salary gap
salary_gap = df.groupby("dept")["salary"].agg(["max", "min"])
salary_gap["gap"] = salary_gap["max"] - salary_gap["min"]
largest_gap_dept = salary_gap["gap"].idxmax()
print("Department with largest salary gap:", largest_gap_dept, "\n")

# 4. Highest paid employee per dept
idx = df.groupby("dept")["salary"].idxmax()
top_earners = df.loc[idx, ["dept", "name", "salary"]]
print("Top earners per dept:\n", top_earners, "\n")

# 5. Build summary DataFrame
summary = df.groupby("dept").agg(
    headcount=("name", "count"),
    avg_salary=("salary", "mean")
).reset_index()

# Merge with top earners
summary = summary.merge(top_earners[["dept", "name"]], on="dept")
summary.rename(columns={"name": "top_earner"}, inplace=True)

print("Summary DataFrame:\n", summary)
```

```python
# 2. Keep only salaries above 70000
above_70k = list(filter(lambda s: s > 70000, salaries))
print("Salaries above 70k:", above_70k)


```

```python
# 3. Formatted strings with zip
formatted = list(map(lambda ns: f"{ns[0]}: Rs.{ns[1]:,}", zip(names, salaries)))
print("Formatted name-salary pairs:", formatted)

```

```python
# 4. Chain: filter > increment > sort
step1 = filter(lambda s: s > 60000, salaries)
step2 = map(lambda s: round(s * 1.05), step1)
step3 = sorted(step2, reverse=True)
print("Filtered, incremented, sorted:", step3)

```

```python
# 5. any() and all()
print("Any salary above 90k?", any(s > 90000 for s in salaries))
print("All salaries above 50k?", all(s > 50000 for s in salaries))
print("All names longer than 3 chars?", all(len(n) > 3 for n in names))
```
